```python
import nltk
from nltk.stem import WordNetLemmatizer

nltk.download("wordnet", quiet=True)
lemmatizer = WordNetLemmatizer()

words = ["running", "better", "cats", "geese"]
for word in words:
  print(f"Word: {word} -> Lemma: {lemmatizer.lemmatize(word, 'v')}")
```
