```python
grammar = {
    "S": [["NP", "VP"]],
    "NP": [["DT", "NN"]],
    "VP": [["VB", "NP"]],
    "DT": [["the"]],
    "NN": [["cat"], ["mat"]],
    "VB": [["chases"]],
}


def top_down_parse(tokens, symbol):
  if not tokens and not symbol:
    return True
  if not symbol or not tokens:
    return False
  current, rest_symbols = symbol[0], symbol[1:]
  if current not in grammar:
    return tokens[0] == current and top_down_parse(tokens[1:], rest_symbols)
  for production in grammar[current]:
    if top_down_parse(tokens, production + rest_symbols):
      return True
  return False


print("Top-down parse:", top_down_parse(["the", "cat", "chases", "the", "mat"], ["S"]))
```
