```python
def transform_tags(tagged_sentence):
  new_tags = []
  for i, (word, tag) in enumerate(tagged_sentence):
    if i > 0 and tagged_sentence[i - 1][1] == "DT" and tag == "NN":
      if word in ["quick", "lazy"]:
        new_tags.append((word, "JJ"))
        continue
    new_tags.append((word, tag))
  return new_tags


print(transform_tags([("the", "DT"), ("quick", "NN"), ("cat", "NN")]))
```
