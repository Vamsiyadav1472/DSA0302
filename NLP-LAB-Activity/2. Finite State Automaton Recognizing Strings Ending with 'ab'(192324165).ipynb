```python
def fsa_ends_with_ab(s):
  state = "S0"
  for char in s:
    if state == "S0":
      state = "S1" if char == "a" else "S0"
    elif state == "S1":
      if char == "b":
        state = "S2"
      elif char == "a":
        state = "S1"
      else:
        state = "S0"
    elif state == "S2":
      state = "S1" if char == "a" else "S0"
  return state == "S2"


print("fsa('cab'):", fsa_ends_with_ab("cab"))
print("fsa('abc'):", fsa_ends_with_ab("abc"))
```
