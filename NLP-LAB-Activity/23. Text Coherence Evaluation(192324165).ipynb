```python
def evaluate_coherence(sent1, sent2):
  w1 = set(sent1.lower().split())
  w2 = set(sent2.lower().split())
  overlap = len(w1.intersection(w2))
  return overlap / max(len(w1), 1)


print(
    "Coherence score:",
    evaluate_coherence("The cat sat on the mat.", "The mat was warm."),
)
```
