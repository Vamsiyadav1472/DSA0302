```python
import nltk
from nltk.wsd import lesk

nltk.download("punkt", quiet=True)
nltk.download("wordnet", quiet=True)

context = "I went to the bank to deposit cash."
print("Lesk definition:", lesk(nltk.word_tokenize(context), "bank").definition())
```
