```python
import re

text = "Contact us at support@example.com or sales@example.org for details."
pattern = r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,}\b"

matches = re.findall(pattern, text)
print("Found emails:", matches)
```
