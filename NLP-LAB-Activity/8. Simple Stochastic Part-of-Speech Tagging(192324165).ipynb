```python
tags_vocab = ["NN", "VB", "DT"]
emission_probs = {"the": {"DT": 0.9}, "cat": {"NN": 0.8}, "sat": {"VB": 0.9}}


def stochastic_tag(sentence):
  tagged = []
  for word in sentence.lower().split():
    best_tag = max(
        tags_vocab, key=lambda t: emission_probs.get(word, {}).get(t, 0.1)
    )
    tagged.append((word, best_tag))
  return tagged


print(stochastic_tag("the cat sat"))
```
