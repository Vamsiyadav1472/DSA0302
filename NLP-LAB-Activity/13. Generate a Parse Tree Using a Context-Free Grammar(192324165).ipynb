```python
import nltk

cfg_grammar = nltk.CFG.fromstring("""
    S -> NP VP
    NP -> DT NN
    VP -> VBD NP
    DT -> 'the'
    NN -> 'cat' | 'dog'
    VBD -> 'chased'
""")
parser = nltk.ChartParser(cfg_grammar)
sentence = "the cat chased the dog".split()
for tree in parser.parse(sentence):
  print(tree)
```
