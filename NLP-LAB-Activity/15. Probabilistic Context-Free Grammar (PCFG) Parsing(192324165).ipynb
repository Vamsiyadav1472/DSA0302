```python
import nltk

pcfg_grammar = nltk.PCFG.fromstring("""
    S -> NP VP [1.0]
    NP -> DT NN [0.5] | 'Alice' [0.5]
    VP -> VBD NP [1.0]
    DT -> 'the' [1.0]
    NN -> 'cat' [1.0]
    VBD -> 'chased' [1.0]
""")
viterbi_parser = nltk.ViterbiParser(pcfg_grammar)
sentence = "Alice chased the cat".split()
for tree in viterbi_parser.parse(sentence):
  print(tree)
```
