```python
def recognize_dialog_act(utterance):
  utterance = utterance.lower()
  if "?" in utterance:
    return "Question"
  elif "hello" in utterance or "hi" in utterance:
    return "Greeting"
  elif "thank" in utterance:
    return "Expression of Gratitude"
  return "Statement"


print(recognize_dialog_act("Can you help me with this?"))
```
