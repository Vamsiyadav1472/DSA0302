```python
import random

corpus = "the cat sat on the mat the dog barked at the cat"
tokens = corpus.split()
bigrams = [(tokens[i], tokens[i + 1]) for i in range(len(tokens) - 1)]

bigram_model = {}
for w1, w2 in bigrams:
  bigram_model.setdefault(w1, []).append(w2)


def generate_text(start_word, length=10):
  current = start_word
  result = [current]
  for _ in range(length - 1):
    if current in bigram_model:
      current = random.choice(bigram_model[current])
      result.append(current)
    else:
      break
  return " ".join(result)


print("Generated:", generate_text("the", 8))
```
