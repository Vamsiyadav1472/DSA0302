```python
import re

patterns = [(r".*ing$", "VBG"), (r".*ed$", "VBD"), (r".*s$", "NNS"), (r".*", "NN")]


def rule_based_tag(word):
  for pattern, tag in patterns:
    if re.match(pattern, word):
      return tag


print([(w, rule_based_tag(w)) for w in ["running", "jumped", "cats", "book"]])
```
