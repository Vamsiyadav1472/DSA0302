```python
def evaluate_fopc(expression, context):
  return eval(expression, {}, context)


ctx = {"John": "John", "Mary": "Mary", "Likes": lambda x, y: f"{x} likes {y}"}
print(evaluate_fopc("Likes('John', 'Mary')", ctx))
```
