```python
import spacy

nlp_spacy = spacy.load("en_core_web_sm")
doc = nlp_spacy("The intelligent student solved the difficult problem.")
for chunk in doc.noun_chunks:
  print(f"Noun Phrase: {chunk.text} | Root: {chunk.root.text}")
```
