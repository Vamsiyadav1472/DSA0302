```python
import re

def simulate_fa(test_string, pattern):
    """Simulates a Finite Automaton using Python's regex engine."""
    return bool(re.fullmatch(pattern, test_string))
```

```python
# Regular expressions for Question 1
regex_1 = {
    "1a_only_0s": r"^0*$",
    "1b_all_binary": r"^[01]*$",
    "1c_except_empty": r"^[01]+$",
    "1d_begins_1_ends_1": r"^1[01]*1\(|^1\)",
    "1e_ends_with_00": r"^[01]*00$",
    "1f_at_least_three_1s": r"^(?:[0]*1){3}[01]*$",
    "1g_three_consecutive_1s": r"^[01]*111[01]*$",
    "1h_contains_110": r"^[01]*110[01]*$",
    "1i_no_110": r"^(?!.*110)[01]*$",
    "1j_two_zeros_not_consecutive": r"^1*01+01*$",
    "1k_third_char_0": r"^[01]{2}0[01]*$",
    "1l_zeros_multiple_of_3": r"^(?:1*01*01*01*)*$",
    "1m_same_start_and_end": r"^(?:0[01]*0|1[01]*1|0|1)$",
    "1n_odd_length": r"^(?:[01]{2})*[01]$",
    "1o_start0_odd_or_start1_even": r"^(?:0(?:[01]{2})*|1(?:[01]{2})+)$",
    "1p_length_1_to_3": r"^[01]{1,3}$"
}

print("--- Testing Part 1 ---")
print("1h ('101101' contains 110):", simulate_fa("101101", regex_1["1h_contains_110"]))
print("1i ('1101' has 110, should be False):", simulate_fa("1101", regex_1["1i_no_110"]))
```

```python
# Question 5: State Transition Tables and FSM Simulator

fsm_diagrams = {
    "Diagram_1": {
        "states": {"q1", "q2", "q3"},
        "start": "q1",
        "accept": {"q1", "q3"},
        "transitions": {
            "q1": {"a": "q1", "b": "q2"},
            "q2": {"a": "q1", "b": "q3"},
            "q3": {"a": "q1", "b": "q3"}
        }
    },
    "Diagram_2": {
        "states": {"q1", "q2", "q3", "q4", "q5", "q6", "q7", "q8"},
        "start": "q1",
        "accept": {"q1", "q3", "q5"},
        "transitions": {
            "q1": {"a": "q2", "b": "q3"},
            "q2": {"a": "q4", "b": "q5"},
            "q3": {"a": "q6", "b": "q7"},
            "q4": {"a": "q8", "b": "q8"},
            "q5": {"a": "q8", "b": "q8"},
            "q6": {"a": "q8", "b": "q8"},
            "q7": {"a": "q8", "b": "q8"},
            "q8": {"a": "q8", "b": "q8"}
        }
    },
    "Diagram_3": {
        "states": {"q1", "q2", "q3"},
        "start": "q1",
        "accept": {"q2"},
        "transitions": {
            "q1": {"a": "q2", "b": "q1"},
            "q2": {"a": "q3", "b": "q2"},
            "q3": {"a": "q2", "b": "q3"}
        }
    },
    "Diagram_4": {
        "states": {"q1", "q2", "q3", "q4"},
        "start": "q1",
        "accept": {"q4"},
        "transitions": {
            "q1": {"a": "q2", "b": "q1"},
            "q2": {"a": "q2", "b": "q3"},
            "q3": {"a": "q4", "b": "q1"},
            "q4": {"a": "q2", "b": "q3"}
        }
    },
    "Diagram_5": {
        "states": {"q1", "q2", "q3", "q4"},
        "start": "q1",
        "accept": {"q3"},
        "transitions": {
            "q1": {"a": "q4", "b": "q2"},
            "q2": {"a": "q3", "b": "q2"},
            "q3": {"a": "q3", "b": "q2"},
            "q4": {"a": "q4", "b": "q4"}
        }
    },
    "Diagram_6": {
        "states": {"q1", "q2", "q3", "q4", "q5", "q6", "q7"},
        "start": "q1",
        "accept": {"q3", "q6"},
        "transitions": {
            "q1": {"a": "q2", "b": "q4"},
            "q2": {"a": "q3", "b": "q5"},
            "q3": {"a": "q3", "b": "q6"},
            "q4": {"a": "q5", "b": "q7"},
            "q5": {"a": "q6", "b": "q7"},
            "q6": {"a": "q6", "b": "q7"},
            "q7": {"a": "q7", "b": "q7"}
        }
    }
}

def simulate_diagram(diag_name, input_string):
    """Simulates a custom FSM using the defined transition table dictionaries."""
    diag = fsm_diagrams[diag_name]
    current_state = diag["start"]

    for char in input_string:
        if char in diag["transitions"][current_state]:
            current_state = diag["transitions"][current_state][char]
        else:
            return False

    return current_state in diag["accept"]

print("--- Question 5: FSM Transition Tables & Simulation ---")
print("Testing Diagram 1 with string 'ab':", simulate_diagram("Diagram_1", "ab"))
print("Testing Diagram 3 with string 'aa':", simulate_diagram("Diagram_3", "aa"))
print("Testing Diagram 4 with string 'aab':", simulate_diagram("Diagram_4", "aab"))
```

```python
# Question 9: Zipf's Law computation

rank_given = 4
freq_given = 600
constant = rank_given * freq_given

target_freq = 300
estimated_rank = constant / target_freq

print("--- Question 9: Zipf's Law ---")
print(f"Given: Rank {rank_given} has frequency {freq_given}")
print(f"Best guess for rank of word with frequency {target_freq} is: {int(estimated_rank)}")
```
