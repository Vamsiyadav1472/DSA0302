1.Python_Regular_Expressions_(Regex)_Script

```python
import re
sample_text = "cau we begin? My mobile number is 9876543210 and my alternative is 1234567890. " \
              "You can reach me at john.doe@example.com or jane_doe123@work.org. " \
              "The event is scheduled for 15-08-2024 and the next one is on 25-12-2024. " \
              "Please Ensure All Capitalized Words Are Found."
print("--- Original Text ---")
print(sample_text, "\n")
date_pattern = r'\b(\d{2})-(\d{2})-(\d{4})\b'
first_date = re.search(date_pattern, sample_text)
if first_date:
    print(f"1. First Date found: {first_date.group()}")
else:
    print("1. No Date found.")
mobile_pattern = r'\b\d{10}\b'
mobile_numbers = re.findall(mobile_pattern, sample_text)
print(f"2. Mobile Numbers found: {mobile_numbers}")
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Z|a-z]{2,7}\b'
emails = re.findall(email_pattern, sample_text)
print(f"3. Email IDs found: {emails}")
cap_word_pattern = r'\b[A-Z][a-z]*\b'
cap_words = re.findall(cap_word_pattern, sample_text)
print(f"4. Capitalized Words found: {cap_words}")
starts_with_cau = bool(re.match(r'^cau\b', sample_text))
print(f"5. Starts with 'cau': {starts_with_cau}")
formatted_text = re.sub(date_pattern, r'\3/\2/\1', sample_text)
print("\n--- Text after changing date format ---")
print(formatted_text)
hidden_phone_text = re.sub(r'\b(\d{6})(\d{4})\b', r'******\2', formatted_text)
print("\n--- Text after hiding phone numbers ---")
print(hidden_phone_text)
```

2.NLTK_Text_Processing_and_Word_Frequency_Script

```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.probability import FreqDist
import string
nltk.download('punkt')
nltk.download('punkt_tab')
paragraph = input("Enter a small paragraph:\n")
sentences = sent_tokenize(paragraph)
print("\n--- Sentences ---")
for i, sent in enumerate(sentences, 1):
    print(f"{i}. {sent}")
words = word_tokenize(paragraph)
cleaned_words = [word.lower() for word in words if word not in string.punctuation]
total_words = len(cleaned_words)
print(f"\nTotal Word Count: {total_words}")
print(f"All Words: {cleaned_words}")
freq_dist = FreqDist(cleaned_words)
most_common_3 = freq_dist.most_common(3)
print("\n--- Top 3 Most Frequent Words ---")
for word, count in most_common_3:
    print(f"{word} : {count}")
```

3.Regular_Expression_Operators_Test_Script

```python
import re

test_cases = [
    {
        "operator": "Optional (?) - matches 0 or 1 occurrence",
        "pattern": r"^colou?r$",
        "strings": ["color", "colour", "colouur", "colr"]
    },
    {
        "operator": "Kleene Star (*) - matches 0 or more occurrences",
        "pattern": r"^ab*c$",
        "strings": ["ac", "abc", "abbc", "xyz"]
    },
    {
        "operator": "Kleene Plus (+) - matches 1 or more occurrences",
        "pattern": r"^ab+c$",
        "strings": ["ac", "abc", "abbc", "abbbc"]
    },
    {
        "operator": "Character Class ([]) - matches any character inside",
        "pattern": r"^[aeiou]$",
        "strings": ["a", "e", "z", "ae"]
    },
    {
        "operator": "Negated Character Class ([^]) - matches anything NOT inside",
        "pattern": r"^[^aeiou]$",
        "strings": ["a", "z", "5", "bz"]
    },
    {
        "operator": "Anchors (^ and \() - start (^) and end (\)) of string",
        "pattern": r"^cat$",
        "strings": ["cat", "cating", "thecat", "cat"]
    }
]

for i, case in enumerate(test_cases, 1):
    print(f"Test {i}: {case['operator']}")
    print(f"Pattern: {case['pattern']}")
    print("-" * 40)

    for s in case["strings"]:

        matched = bool(re.search(case["pattern"], s)) and (bool(re.match(case["pattern"], s)) if '^' in case["pattern"] and '$' in case["pattern"] else True)

        is_match = bool(re.fullmatch(case["pattern"], s)) if (case["pattern"].startswith("^") and case["pattern"].endswith("$")) else bool(re.search(case["pattern"], s))

        result_str = "Yes" if is_match else "No"
        print(f"  String '{s}' -> {result_str}")
    print("\n" + "=" * 40 + "\n")
```

4.Overview of Regex Operators

```python
import re
pattern = r'^[A-Z][^0-9]+[a-z]?$'
re_compiled = re.compile(pattern)

test_strings = [
    "Apple",   # Valid: Starts with A, followed by non-digits, optional lowercase e at end
    "A123",    # Invalid: Contains digits where non-digits are required ([^0-9]+)
    "apple",   # Invalid: Does not start with an uppercase letter
    "Test!",   # Valid: Starts with T, followed by non-digits, optional lowercase character is absent
    "H",       # Invalid: Missing the one-or-more non-digit characters after the first letter
]

print(f"Testing strings against pattern: {pattern}")
print("----------------------------------------")
for s in test_strings:
    if re_compiled.match(s):
        print(f"String: {s:<8} -> yes")
    else:
        print(f"String: {s:<8} -> no")
```

5.Finite State Automaton Pattern Recognizer

```python

def simulate_dfa(input_str):
    current_state = "S0"
    path = [current_state]

    for ch in input_str:
        if current_state == "S0":
            if ch == 'a':
                current_state = "S1"
            elif ch == 'b':
                current_state = "S0"
            else:
                return path, "Reject (Invalid Character)"
        elif current_state == "S1":
            if ch == 'a':
                current_state = "S1"
            elif ch == 'b':
                current_state = "S2"
            else:
                return path, "Reject (Invalid Character)"
        elif current_state == "S2":
            if ch == 'a':
                current_state = "S1"
            elif ch == 'b':
                current_state = "S0"
            else:
                return path, "Reject (Invalid Character)"
        path.append(current_state)

    if current_state == "S2":
        return path, "Accept"
    return path, "Reject"

test_strings = [
    "ab",
    "aab",
    "aba",
    "bbab",
    "bba",
]

print("DFA Pattern Recognizer (Target pattern ending: 'ab')")
print("-----------------------------------------------------")

for s in test_strings:
    path, result = simulate_dfa(s)
    print(f"Input: {s:<6} | States: {path} | Result: {result}")
```

6.DFA_State_Traversal_&_Validation_Script_Pyt

```python
class DFAEndingWithAB:
    def __init__(self):

        self.states = {"Q0", "Q1", "Q2"}
        self.start_state = "Q0"
        self.accept_states = {"Q2"}

    def run(self, input_string):
        current_state = self.start_state
        path = [current_state]

        print(f"--- Processing Input: '{input_string}' ---")
        print(f"Start State: {current_state}")

        for char in input_string:
            if char not in {'a', 'b'}:
                print(f"Error: Invalid character '{char}' in input string.")
                return False

            # Transition Logic
            if current_state == "Q0":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q0"

            elif current_state == "Q1":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q2"

            elif current_state == "Q2":
                if char == 'a':
                    current_state = "Q1"
                else:  # char == 'b'
                    current_state = "Q0"

            path.append(current_state)
            print(f"Read '{char}' -> Transitioned to state {current_state}")

        is_accepted = current_state in self.accept_states
        print(f"Path of States: {' -> '.join(path)}")
        if is_accepted:
            print("Result: Accepted\n")
        else:
            print("Result: Rejected\n")

        return is_accepted

dfa = DFAEndingWithAB()
test_strings = ["ab", "aab", "baba", "aba", "bb", "a", "b"]

for s in test_strings:
    dfa.run(s)
```

8.Morphological Analysis Using NLTK

```python
import nltk
from nltk.stem import WordNetLemmatizer
from nltk.corpus import wordnet

# Ensure necessary NLTK data packages are available
for resource in ['punkt', 'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng', 'wordnet', 'omw-1.4', 'punkt_tab']:
    try:
        nltk.data.find(f'tokenizers/{resource}' if 'punkt' in resource else f'taggers/{resource}' if 'tagger' in resource else f'corpora/{resource}')
    except LookupError:
        nltk.download(resource, quiet=True)

def get_wordnet_pos(treebank_tag):
    """Map POS tag to first character lemmatize() accepts"""
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

def analyze_morphology(sentence):
    lemmatizer = WordNetLemmatizer()

    # Tokenize and perform Part of Speech tagging
    tokens = nltk.word_tokenize(sentence)
    pos_tags = nltk.pos_tag(tokens)

    print(f"Analyzing Sentence: \"{sentence}\"")
    print(f"{'Word':<12} | {'POS Tag':<10} | {'Root (Lemma)':<15} | {'Grammatical Features'}")
    print("-" * 65)

    for word, tag in pos_tags:
        # Determine WordNet POS and get lemma (root)
        wn_pos = get_wordnet_pos(tag)
        lemma = lemmatizer.lemmatize(word.lower(), pos=wn_pos)

        # Derive basic grammatical features based on POS tag
        features = "Standard"
        if tag.startswith('N'):
            features = "Plural Noun" if tag == 'NNS' or tag == 'NNPS' else "Singular Noun"
        elif tag.startswith('V'):
            if tag == 'VBD': features = "Verb, past tense"
            elif tag == 'VBG': features = "Verb, gerund/present participle"
            elif tag == 'VBN': features = "Verb, past participle"
            elif tag == 'VBP': features = "Verb, non-3rd person singular present"
            elif tag == 'VBZ': features = "Verb, 3rd person singular present"
            elif tag == 'VB':  features = "Verb, base form"
        elif tag.startswith('J'):
            features = "Adjective"

        print(f"{word:<12} | {tag:<10} | {lemma:<15} | {features}")

if __name__ == "__main__":
    sample_sentence = "The running children happily played outside."
    analyze_morphology(sample_sentence)
```

9.Porter Stemming and Lemmatization Comparison

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Ensure required NLTK data is available
for resource in ['wordnet', 'omw-1.4']:
    try:
        nltk.data.find(f'corpora/{resource}')
    except LookupError:
        nltk.download(resource, quiet=True)

def compare_stemming_and_lemmatization():
    # Initialize the Porter Stemmer and WordNet Lemmatizer
    porter = PorterStemmer()
    lemmatizer = WordNetLemmatizer()

    # Sample list of words with various inflections
    sample_words = [
        "running", "runs", "ran",
        "easily", "fairly",
        "cats", "better",
        "studying", "studies", "computed"
    ]

    print("Comparison: Porter Stemmer vs. Lemmatization")
    print(f"{'Original Word':<15} | {'Porter Stem':<15} | {'Lemma (Root)':<15}")
    print("-" * 51)

    for word in sample_words:
        # Apply Porter Stemmer (chops suffixes based on rules)
        stem = porter.stem(word)

        # Apply WordNet Lemmatizer (finds valid dictionary root form)
        # Note: lemmatize defaults to noun, but works well for many words here
        lemma = lemmatizer.lemmatize(word)

        print(f"{word:<15} | {stem:<15} | {lemma:<15}")

if __name__ == "__main__":
    compare_stemming_and_lemmatization()
```
