Problem 1: Tokenization with a Regular Expression

```python
import re
import sys
def tokenize_text(lines):
    token_pattern = re.compile(r"[A-Za-z]+(?:['-][A-Za-z]+)*")
    output_lines = []
    for line in lines:
        tokens = token_pattern.findall(line)
        count = len(tokens)
        if count > 0:
            joined_tokens = " ".join(tokens)
            output_lines.append(f"{count} {joined_tokens}")
        else:
            output_lines.append("0")
    return output_lines
def solve_problem_1():
    input_data = sys.stdin.read().splitlines()
    if not input_data:
        return
    try:
        Q = int(input_data[0].strip())
        lines_to_process = [line.strip() for line in input_data[1:Q+1]]
    except ValueError:
        lines_to_process = [line.strip() for line in input_data]
    for result in tokenize_text(lines_to_process):
        print(result)
if __name__ == "__main__":
    sample_input_lines = [
        "The dog's tail wagged -- twice!",
        "state-of-the-art NLP systems, circa 2026.",
        "12345 ... ???",
        "Don't re-use rock--roll O'Neill's x"
    ]
    print("--- Sample Output ---")
    for result in tokenize_text(sample_input_lines):
        print(result)
```

Problem 2: Simulating a Finite State Automaton

```python
import sys
def simulate_automaton(input_lines):
  idx = 0
  n, k = map(int, input_lines[idx].split())
  idx += 1
  alphabet = input_lines[idx].split()
  idx += 1
  start_state = int(input_lines[idx].strip())
  idx += 1
  parts = input_lines[idx].split()
  m_acc = int(parts[0])
  accepting_states = set(map(int, parts[1:]))
  idx += 1
  t = int(input_lines[idx].strip())
  idx += 1
  transitions = []
  for _ in range(t):
    p, c, r = input_lines[idx].split()
    transitions.append((int(p), c, int(r)))
    idx += 1
  Q = int(input_lines[idx].strip())
  idx += 1
  queries = []
  for _ in range(Q):
    queries.append(input_lines[idx].strip())
    idx += 1
  output_results = []
  for query in queries:
    if query == "#":
      symbols = []
    else:
      symbols = list(query)
    current_states = {start_state}
    for symbol in symbols:
      next_states = set()
      for state in current_states:
        for p, c, r in transitions:
          if p == state and c == symbol:
            next_states.add(r)
      current_states = next_states
      if not current_states:
        break
    if current_states & accepting_states:
      output_results.append("ACCEPT")
    else:
      output_results.append("REJECT")
  return output_results
def solve_problem_2():
  input_data = sys.stdin.read().splitlines()
  if not input_data:
    return
  for result in simulate_automaton(input_data):
    print(result)
if __name__ == "__main__":
  sample_input_lines = [
      "5 3",
      "b a !",
      "0",
      "1 4",
      "5",
      "0 b 1",
      "1 a 2",
      "2 a 3",
      "3 a 3",
      "3 ! 4",
      "5",
      "baa!",
      "baaa!",
      "ba!",
      "b",
      "baa",
  ]
  print("--- Sample Output ---")
  for result in simulate_automaton(sample_input_lines):
    print(result)
```

Problem 3: Inflectional Morphology Generator

```python
import sys
def is_vowel(char):
  return char in "aeiou"
def count_vowel_groups(stem):
  groups = 0
  in_vowel = False
  for char in stem:
    if is_vowel(char):
      if not in_vowel:
        groups += 1
        in_vowel = True
    else:
      in_vowel = False
  return groups
def is_doublable(stem):
  if len(stem) < 3:
    return False
  c1, v1, c2 = stem[-3], stem[-2], stem[-1]
  if is_vowel(c1) or not is_vowel(v1) or is_vowel(c2):
    return False
  if c2 in "wxy":
    return False
  if count_vowel_groups(stem) != 1:
    return False
  return True
def apply_inflection(stem, feature):
  if feature in ("N+PL", "V+3SG"):
    if stem.endswith(("s", "x", "z", "ch", "sh")):
      return stem + "es"
    elif (
        len(stem) >= 2
        and not is_vowel(stem[-1])
        and stem[-1] == "o"
        and not is_vowel(stem[-2])
    ):
      return stem + "es"
    elif len(stem) >= 2 and stem[-1] == "y" and not is_vowel(stem[-2]):
      return stem[:-1] + "ies"
    else:
      return stem + "s"
  elif feature == "V+PROG":
    if len(stem) >= 3 and stem[-1] == "e" and not is_vowel(stem[-2]):
      return stem[:-1] + "ing"
    elif is_doublable(stem):
      return stem + stem[-1] + "ing"
    else:
      return stem + "ing"
  elif feature in ("V+PAST", "ADJ+COMP", "ADJ+SUP"):
    suffix_map = {
        "V+PAST": ("d", "ed"),
        "ADJ+COMP": ("r", "er"),
        "ADJ+SUP": ("st", "est"),
    }
    sfx_short, sfx_long = suffix_map[feature]
    if stem.endswith("e"):
      return stem + sfx_short
    elif len(stem) >= 2 and stem[-1] == "y" and not is_vowel(stem[-2]):
      return stem[:-1] + "i" + sfx_long
    elif is_doublable(stem):
      return stem + stem[-1] + sfx_long
    else:
      return stem + sfx_long
  return stem
def process_morphology(input_lines):
  idx = 0
  L = int(input_lines[idx].strip())
  idx += 1
  lexicon = {}
  for _ in range(L):
    stem, feature, form = input_lines[idx].split()
    lexicon[(stem, feature)] = form
    idx += 1
  Q = int(input_lines[idx].strip())
  idx += 1
  output_results = []
  for _ in range(Q):
    stem, feature = input_lines[idx].split()
    idx += 1
    if (stem, feature) in lexicon:
      output_results.append(lexicon[(stem, feature)])
    else:
      output_results.append(apply_inflection(stem, feature))
  return output_results
def solve_problem_3():
  input_data = sys.stdin.read().splitlines()
  if not input_data:
    return
  for result in process_morphology(input_data):
    print(result)
if __name__ == "__main__":
  sample_input_lines = [
      "3",
      "goose N+PL geese",
      "go V+PAST went",
      "good ADJ+COMP better",
      "8",
      "cat N+PL",
      "church N+PL",
      "baby N+PL",
      "stop V+PROG",
      "make V+PROG",
      "carry V+PAST",
      "goose N+PL",
      "go V+PAST",
  ]
  print("--- Sample Output ---")
  for result in process_morphology(sample_input_lines):
    print(result)
```

Problem 4: Finite-State Morphological Parsing

```python
import sys
def is_vowel(char):
  return char in "aeiou"
def count_vowel_groups(stem):
  groups = 0
  in_vowel = False
  for char in stem:
    if is_vowel(char):
      if not in_vowel:
        groups += 1
        in_vowel = True
    else:
      in_vowel = False
  return groups
def is_doublable(stem):
  if len(stem) < 3:
    return False
  c1, v1, c2 = stem[-3], stem[-2], stem[-1]
  if is_vowel(c1) or not is_vowel(v1) or is_vowel(c2):
    return False
  if c2 in "wxy":
    return False
  if count_vowel_groups(stem) != 1:
    return False
  return True
def apply_inflection(stem, feature):
  if feature in ("N+PL", "V+3SG"):
    if stem.endswith(("s", "x", "z", "ch", "sh")):
      return stem + "es"
    elif (
        len(stem) >= 2
        and not is_vowel(stem[-1])
        and stem[-1] == "o"
        and not is_vowel(stem[-2])
    ):
      return stem + "es"
    elif len(stem) >= 2 and stem[-1] == "y" and not is_vowel(stem[-2]):
      return stem[:-1] + "ies"
    else:
      return stem + "s"
  elif feature == "V+PROG":
    if len(stem) >= 3 and stem[-1] == "e" and not is_vowel(stem[-2]):
      return stem[:-1] + "ing"
    elif is_doublable(stem):
      return stem + stem[-1] + "ing"
    else:
      return stem + "ing"
  elif feature in ("V+PAST", "ADJ+COMP", "ADJ+SUP"):
    suffix_map = {
        "V+PAST": ("d", "ed"),
        "ADJ+COMP": ("r", "er"),
        "ADJ+SUP": ("st", "est"),
    }
    sfx_short, sfx_long = suffix_map[feature]
    if stem.endswith("e"):
      return stem + sfx_short
    elif len(stem) >= 2 and stem[-1] == "y" and not is_vowel(stem[-2]):
      return stem[:-1] + "i" + sfx_long
    elif is_doublable(stem):
      return stem + stem[-1] + sfx_long
    else:
      return stem + sfx_long
  return stem
def parse_morphology(input_lines):
  idx = 0
  S = int(input_lines[idx].strip())
  idx += 1
  lexicon_entries = []
  for _ in range(S):
    stem, pos = input_lines[idx].split()
    lexicon_entries.append((stem, pos))
    idx += 1
  Q = int(input_lines[idx].strip())
  idx += 1
  queries = []
  for _ in range(Q):
    queries.append(input_lines[idx].strip())
    idx += 1
  features_map = {
      "N": ["PL"],
      "V": ["3SG", "PROG", "PAST"],
      "ADJ": ["COMP", "SUP"],
  }
  output_results = []
  for query in queries:
    valid_analyses = set()
    for stem, pos in lexicon_entries:
      if query == stem:
        valid_analyses.add(f"{stem}+{pos}")
      for feat in features_map[pos]:
        if pos == "N" and feat == "PL":
          p3_feat = "N+PL"
        elif pos == "V" and feat == "3SG":
          p3_feat = "V+3SG"
        elif pos == "V" and feat == "PROG":
          p3_feat = "V+PROG"
        elif pos == "V" and feat == "PAST":
          p3_feat = "V+PAST"
        elif pos == "ADJ" and feat == "COMP":
          p3_feat = "ADJ+COMP"
        elif pos == "ADJ" and feat == "SUP":
          p3_feat = "ADJ+SUP"
        else:
          continue
        generated = apply_inflection(stem, p3_feat)
        if generated == query:
          valid_analyses.add(f"{stem}+{pos}+{feat}")
    if valid_analyses:
      sorted_analyses = sorted(list(valid_analyses))
      output_results.append(", ".join(sorted_analyses))
    else:
      output_results.append("UNKNOWN")
  return output_results
def solve_problem_4():
  input_data = sys.stdin.read().splitlines()
  if not input_data:
    return
  for result in parse_morphology(input_data):
    print(result)
if __name__ == "__main__":
  sample_input_lines = [
      "5",
      "cat N",
      "watch N",
      "watch V",
      "study V",
      "big ADJ",
      "8",
      "cats",
      "watches",
      "watching",
      "studied",
      "bigger",
      "cat",
      "biggest",
      "elephants",
  ]
  print("--- Sample Output ---")
  for result in parse_morphology(sample_input_lines):
    print(result)
```

Problem 5: The Porter Stemmer

```python
import sys
def porter_is_consonant(word, i):
  char = word[i]
  if char in "aeiou":
    return False
  if char == "y":
    if i == 0:
      return True
    else:
      return not porter_is_consonant(word, i - 1)
  return True
def get_porter_measure(stem):
  n = len(stem)
  i = 0
  while i < n and porter_is_consonant(stem, i):
    i += 1
  m = 0
  while True:
    v_start = i
    while i < n and not porter_is_consonant(stem, i):
      i += 1
    if i == v_start:
      break
    c_start = i
    while i < n and porter_is_consonant(stem, i):
      i += 1
    if i == c_start:
      break
    m += 1
  return m
def porter_has_vowel(stem):
  for i in range(len(stem)):
    if not porter_is_consonant(stem, i):
      return True
  return False
def porter_ends_double_consonant(stem):
  if (
      len(stem) >= 2
      and stem[-1] == stem[-2]
      and porter_is_consonant(stem, len(stem) - 1)
  ):
    return True
  return False
def porter_ends_cvc(stem):
  if len(stem) >= 3:
    c1 = porter_is_consonant(stem, len(stem) - 3)
    v = not porter_is_consonant(stem, len(stem) - 2)
    c2 = porter_is_consonant(stem, len(stem) - 1)
    if c1 and v and c2 and stem[-1] not in "wxy":
      return True
  return False
def porter_step_1(word):
  if word.endswith("sses"):
    word = word[:-2]
  elif word.endswith("ies"):
    word = word[:-2]
  elif word.endswith("ss"):
    pass
  elif word.endswith("s"):
    word = word[:-1]
  ed_ing_removed = False
  if word.endswith("eed"):
    stem = word[:-3]
    if get_porter_measure(stem) > 0:
      word = stem + "ee"
  elif word.endswith("ed"):
    stem = word[:-2]
    if porter_has_vowel(stem):
      word = stem
      ed_ing_removed = True
  elif word.endswith("ing"):
    stem = word[:-3]
    if porter_has_vowel(stem):
      word = stem
      ed_ing_removed = True
  if ed_ing_removed:
    if word.endswith("at") or word.endswith("bl") or word.endswith("iz"):
      word = word + "e"
    elif porter_ends_double_consonant(word) and word[-1] not in "lsz":
      word = word[:-1]
    elif get_porter_measure(word) == 1 and porter_ends_cvc(word):
      word = word + "e"
  if word.endswith("y"):
    stem = word[:-1]
    if porter_has_vowel(stem):
      word = stem + "i"
  return word
def process_porter(input_lines):
  Q = int(input_lines[0].strip())
  output_results = []
  for i in range(1, Q + 1):
    if i < len(input_lines):
      word = input_lines[i].strip()
      output_results.append(porter_step_1(word))
  return output_results
def solve_problem_5():
  input_data = sys.stdin.read().splitlines()
  if not input_data:
    return
  for result in process_porter(input_data):
    print(result)
if __name__ == "__main__":
  sample_input_lines = [
      "8",
      "caresses",
      "ponies",
      "cats",
      "feed",
      "agreed",
      "motoring",
      "hopping",
      "happy",
  ]
  print("--- Sample Output ---")
  for result in process_porter(sample_input_lines):
    print(result)
```
