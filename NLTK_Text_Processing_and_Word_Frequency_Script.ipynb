```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.probability import FreqDist
import string
nltk.download('punkt')
nltk.download('punkt_tab')
paragraph = input("Enter a small paragraph:\n")
sentences = sent_tokenize(paragraph)
print("\n--- Sentences ---")
for i, sent in enumerate(sentences, 1):
    print(f"{i}. {sent}")
words = word_tokenize(paragraph)
cleaned_words = [word.lower() for word in words if word not in string.punctuation]
total_words = len(cleaned_words)
print(f"\nTotal Word Count: {total_words}")
print(f"All Words: {cleaned_words}")
freq_dist = FreqDist(cleaned_words)
most_common_3 = freq_dist.most_common(3)
print("\n--- Top 3 Most Frequent Words ---")
for word, count in most_common_3:
    print(f"{word} : {count}")
```
