Code 2: Top-Down Parser Implementation

```python
import nltk
from nltk import CFG
from nltk.parse.earleychart import EarleyChartParser

# 1. Define a clean grammar fragment (free of infinite left-recursion loops)
helpdesk_grammar = CFG.fromstring("""
    S -> NP VP | IMP_VP

    NP -> Det N | ProperNoun
    VP -> V NP | V NP PP
    IMP_VP -> V NP | V NP PP
    PP -> P NP

    Det -> 'the' | 'a' | 'an'
    ProperNoun -> 'Vellore' | 'Chennai' | 'Library'
    N -> 'student' | 'room' | 'seminar' | 'fee'
    V -> 'book' | 'show' | 'submit'
    P -> 'near' | 'for' | 'with'
""")

# 2. Initialize the Earley Chart Parser (handles ambiguity & left-recursion)
parser = EarleyChartParser(helpdesk_grammar)

# 3. Test tokens (must match lowercase terminals in the grammar)
tokens = ['book', 'the', 'room']

# 4. Parse and display output trees
trees = list(parser.parse(tokens))

if trees:
    print(f"Found {len(trees)} parse tree(s):\n")
    for tree in trees:
        print(tree)
        tree.pretty_print()
else:
    print("No parse found. Check token spelling or grammar rules.")
```

Code 3: Earley Chart Parser Implementation

```python
import nltk
from nltk import CFG
from nltk.parse.earleychart import EarleyChartParser

# 1. Define a CFG grammar fragment covering helpdesk queries
helpdesk_grammar = CFG.fromstring("""
    S -> NP VP | IMP_VP

    NP -> Det N
    VP -> V NP
    IMP_VP -> V NP

    Det -> 'the' | 'a'
    N -> 'form' | 'student' | 'notice'
    V -> 'submit' | 'show' | 'book'
""")

def run_earley_parser(grammar, tokens):
    # Initialize the Earley Chart Parser with trace=0 (silent) or trace=1 (debug info)
    parser = EarleyChartParser(grammar, trace=0)
    return list(parser.parse(tokens))

# 2. Test sentence tokens matching the lexicon
tokens = ['submit', 'the', 'form']

# 3. Execute the parser
trees = run_earley_parser(helpdesk_grammar, tokens)

# 4. Print output and inspect trees
if trees:
    print(f"Success! Found {len(trees)} parse tree(s):\n")
    for tree in trees:
        print(tree)
        tree.pretty_print()
else:
    print("No parse found. Check if all tokens are defined in your grammar's lexicon.")
```
