```python
!pip install pyspellchecker
from spellchecker import SpellChecker
import string
def check_spelling():
    spell = SpellChecker()
    paragraph = input("Enter a paragraph:\n")
    target_word = input("\nEnter a word to check:\n").strip().lower()
    translator = str.maketrans('', '', string.punctuation)
    cleaned_words = paragraph.translate(translator).lower().split()
    misspelled = spell.unknown(cleaned_words)
    print("\n--- Paragraph Spell Check ---")
    if not misspelled:
        print("No misspelled words found in the paragraph.")
    else:
        print("Misspelled words found (with top suggestions):")
        for word in misspelled:
            print(f" - '{word}' -> Did you mean '{spell.correction(word)}'?")
    print("\n--- Target Word Check ---")
    if target_word in misspelled:
        print(f"The target word '{target_word}' is MISSPELLED.")
        print(f"Correction suggestion: '{spell.correction(target_word)}'")
        print(f"Other possible options: {spell.candidates(target_word)}")
    elif target_word in cleaned_words:
        print(f"The target word '{target_word}' is in the paragraph and is spelled correctly.")
    else:
        if spell.unknown([target_word]):
            print(f"The word '{target_word}' was not found in the paragraph, and it is MISSPELLED.")
            print(f"Correction suggestion: '{spell.correction(target_word)}'")
        else:
            print(f"The word '{target_word}' was not found in the paragraph, but it is a valid word.")
check_spelling()
```
